```python
#Q3

import pandas as pd

# Create DataFrame
data = {
    "Name": ["Amit", "Neha", "Rahul", "Sneha"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 88]
}

df = pd.DataFrame(data)

# Display first 2 rows
print("First 2 rows:")
print(df.head(2))

# Display only Name and Marks columns
print("\nName and Marks columns:")
print(df[["Name", "Marks"]])

# Add a new column named "Result"
df["Result"] = ["Pass", "Pass", "Pass", "Pass"]

print("\nDataFrame with Result column:")
print(df)

# Display students with marks greater than 80
print("\nStudents with marks > 80:")
print(df[df["Marks"] > 80])
```

```python
#Q4

import pandas as pd
import numpy as np

# Create DataFrame
data = {
    "Name": ["A", "B", "C", "D"],
    "Age": [25, np.nan, 27, 24],
    "Salary": [30000, 40000, np.nan, 35000]
}

df = pd.DataFrame(data)

# Identify missing values
print("Missing values:")
print(df.isnull())

# Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# Display cleaned dataset
print("\nCleaned Dataset:")
print(df)
```

```python
#Q5

import pandas as pd

# Load CSV file
df = pd.read_csv("Iris.csv")

# Display dataset information
print("Dataset Information:")
print(df.info())

# Find total rows and columns
print("\nTotal Rows and Columns:")
print(df.shape)      # (rows, columns)

# Find average sepal length
print("\nAverage Sepal Length:")
print(df["SepalLengthCm"].mean())

# Find different species names
print("\nDifferent Species Names:")
print(df["Species"].unique())

# Display summary statistics
print("\nSummary Statistics:")
print(df.describe())
```
