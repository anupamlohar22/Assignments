```python
#Q1

def student(**kwargs):
    for k, v in kwargs.items():
        print(k, ":", v)

student(Name="Anupam", Age=20, Course="BCA")
```

```python
#Q2

def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

```python
#Q4

import json

data = {"name": "Anupam", "age": 20}

with open("data.json", "w") as f:
    json.dump(data, f)

with open("data.json", "r") as f:
    print(json.load(f))
```
