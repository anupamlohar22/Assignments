## Q1. Explain Support Vector Machine and its working principle.

Support Vector Machine (SVM) is a supervised machine learning algorithm used mainly for classification and sometimes for regression. It finds the optimal boundary (hyperplane) that separates different classes with the maximum possible margin.

**Working Principle:**
1. Represent data points in an n-dimensional space.
2. Find the hyperplane that best separates the classes.
3. Maximize the distance (margin) between the hyperplane and the nearest data points.
4. Use support vectors to define the decision boundary.
5. Predict the class of new observations based on their position relative to the hyperplane.


## Q2. What are support vectors and hyperplanes in SVM?

**Support Vectors:**  
Support vectors are the data points closest to the decision boundary. They determine the position and orientation of the hyperplane.

**Hyperplane:**  
A hyperplane is the boundary that separates different classes in the dataset.

**Example:**  
If students are classified as Pass and Fail using study hours and attendance, the separating line between both groups in a two-dimensional graph is called the hyperplane, while the nearest students to that line are the support vectors.


## Q3. Differentiate between Linear SVM, Non-Linear SVM, and Kernel Functions

| Feature | Linear SVM | Non-Linear SVM | Kernel Functions |
|---|---|---|---|
| Data Type | Linearly separable | Non-linearly separable | Transform data into higher dimensions |
| Decision Boundary | Straight line/plane | Curved boundary | Enables non-linear separation |
| Complexity | Lower | Higher | Depends on kernel used |
| Examples | Spam detection | Image recognition | RBF, Polynomial, Sigmoid |

**Common Kernel Functions:**
- Linear Kernel
- Polynomial Kernel
- Radial Basis Function (RBF) Kernel
- Sigmoid Kernel


## Q4. SVM Classifier for Student Result Prediction

```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

data = pd.DataFrame({
    'Study_Hours':[1,2,3,4,5,6],
    'Attendance':[50,55,60,70,75,85],
    'Result':[0,0,0,1,1,1]
})

X = data[['Study_Hours','Attendance']]
y = data['Result']

model = SVC(kernel='linear')
model.fit(X, y)

prediction = model.predict([[4.5,72]])[0]
accuracy = accuracy_score(y, model.predict(X))

print("Predicted Result:", prediction)
print("Accuracy Score:", accuracy)
```

## Q5. SVM Model for Fruit Classification

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, classification_report

data = pd.DataFrame({
    'Weight':[120,130,150,170,180,200],
    'Size':[6,6.5,7,8,8.5,9],
    'Fruit':['Apple','Apple','Apple','Orange','Orange','Orange']
})

le = LabelEncoder()
data['Fruit_Encoded'] = le.fit_transform(data['Fruit'])

X = data[['Weight','Size']]
y = data['Fruit_Encoded']

model = SVC(kernel='linear')
model.fit(X, y)

prediction = model.predict([[160,7.5]])
predicted_fruit = le.inverse_transform(prediction)[0]

y_pred = model.predict(X)

print("Predicted Fruit:", predicted_fruit)
print("\nConfusion Matrix:")
print(confusion_matrix(y, y_pred))
print("\nClassification Report:")
print(classification_report(y, y_pred, target_names=le.classes_))
```
