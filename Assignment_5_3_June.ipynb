```python
# Peterson Number

def factorial(n):
    f = 1
    for i in range(1, n + 1):
        f *= i
    return f

num = int(input("Enter a number: "))
temp = num
sum_fact = 0

while temp > 0:
    digit = temp % 10
    sum_fact += factorial(digit)
    temp //= 10

if sum_fact == num:
    print("Peterson Number")
else:
    print("Not a Peterson Number")
```

```python
# Circular Prime Number

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

num = input("Enter a number: ")
flag = True

for i in range(len(num)):
    rotated = int(num[i:] + num[:i])
    if not is_prime(rotated):
        flag = False
        break

if flag:
    print("Circular Prime Number")
else:
    print("Not a Circular Prime Number")
```

```python
# Harshad Number

num = int(input("Enter a number: "))
digit_sum = sum(int(d) for d in str(num))

if num % digit_sum == 0:
    print("Harshad Number")
else:
    print("Not a Harshad Number")
```

```python
# Magic Number

num = int(input("Enter a number: "))

while num > 9:
    s = 0
    while num > 0:
        s += num % 10
        num //= 10
    num = s

if num == 1:
    print("Magic Number")
else:
    print("Not a Magic Number")
```

```python
# Generate Pascal Triangle

n = int(input("Enter number of rows: "))

for i in range(n):
    num = 1
    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
    print()
```

```python
# Duck Number

num = input("Enter a number: ")

if num[0] != '0' and '0' in num:
    print("Duck Number")
else:
    print("Not a Duck Number")
```
