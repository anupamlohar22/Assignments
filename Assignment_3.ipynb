```python
#Q1

price = int(input("enter the price: "))
quantity = int(input("enter the qty: "))
discount = int(input("enter the discount: "))
def calculate_bill(price,quantity,discount):
    final_bill = ((price*quantity)-(price*quantity*(discount/100)))
    print("Final Bill:",final_bill)
calculate_bill(price,quantity,discount)
```

```python
#Q2

basic = int(input("enter basic salary: "))
def calculate_salary(basic,bonus=2000,tax=10):
    print("net salary:",basic+bonus-((basic+bonus)*(tax/100)))

calculate_salary(basic)

bonus = int(input("enter the bonus to calculate another salary: "))
tax = int(input("enter the tax to calculate another salary: "))
calculate_salary(basic,bonus,tax)
```

```python
#Q3

def find_max(*numbers):
    largest=numbers[0]
    for i in range(len(numbers)):
        current=numbers[i]
        if current>largest:
            largest = current
        else:
            continue
    print("largest:",largest)

find_max(10,7,20,30,50,7)
```

```python
#Q4

def student_details(**data):
    for key,value in data.items():
        print(f'{key}:{value}')
student_details(name="anupam",age=21,course="python",city="jaipur")
```

```python
#Q5

n = int(input("enter n: "))
def fibonacci(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return fibonacci(n-1)+fibonacci(n-2) 
for i in range(0,n):
    print(fibonacci(i),end=" ")
```
