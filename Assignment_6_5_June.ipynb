```python
#Q1

# Create a class named Employee
class Employee:

    # Constructor to initialize employee details
    def __init__(self, emp_name, emp_id, salary):
        self.emp_name = emp_name   # Store employee name
        self.emp_id = emp_id       # Store employee ID
        self.salary = salary       # Store employee salary

    # Method to display employee details
    def display(self):
        print("Employee Name:", self.emp_name)  # Print name
        print("Employee ID:", self.emp_id)      # Print ID
        print("Salary:", self.salary)           # Print salary
        print()                                 # Print blank line


# Create first employee object
emp1 = Employee("Anupam", 101, 50000)

# Create second employee object
emp2 = Employee("Rahul", 102, 60000)

# Create third employee object
emp3 = Employee("Priya", 103, 55000)

# Display details of first employee
emp1.display()

# Display details of second employee
emp2.display()

# Display details of third employee
emp3.display()
```

```python
#Q2

# Create a class named Calculator
class Calculator:

    # Method to add two numbers
    def add(self, a, b):
        # Return the sum of a and b
        return a + b

    # Method to subtract two numbers
    def subtract(self, a, b):
        # Return the difference of a and b
        return a - b

    # Method to multiply two numbers
    def multiply(self, a, b):
        # Return the product of a and b
        return a * b

    # Method to divide two numbers
    def divide(self, a, b):
        # Check if the divisor is zero
        if b == 0:
            # Return an error message if division by zero is attempted
            return "Error: Division by zero is not allowed"

        # Return the quotient of a and b
        return a / b


# Create an object of the Calculator class
calc = Calculator()

# Call the add method and print the result
print("Addition:", calc.add(10, 5))

# Call the subtract method and print the result
print("Subtraction:", calc.subtract(10, 5))

# Call the multiply method and print the result
print("Multiplication:", calc.multiply(10, 5))

# Call the divide method and print the result
print("Division:", calc.divide(10, 5))
```

```python
#Q3

# Create a class named Book
class Book:

    # Constructor to initialize title, author, and price
    def __init__(self, title, author, price):
        # Store the title in the object
        self.title = title

        # Store the author in the object
        self.author = author

        # Store the price in the object
        self.price = price

    # Method to display book details
    def display_details(self):
        # Print the title of the book
        print("Title:", self.title)

        # Print the author of the book
        print("Author:", self.author)

        # Print the price of the book
        print("Price:", self.price)


# Create an object of the Book class
book1 = Book("Python Basics", "John Smith", 499)

# Call the method to print book details
book1.display_details()
```

```python
#Q4

# Create a parent class named Vehicle
class Vehicle:

    # Define a method to display a message
    def display(self):
        print("This is a Vehicle")


# Create a child class named Car that inherits Vehicle
class Car(Vehicle):

    # Define a method to call the parent class method
    def show(self):
        # Call the display() method of the parent class
        super().display()


# Create an object of the Car class
c = Car()

# Call the child class method
c.show()
```
