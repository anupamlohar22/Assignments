## Q1. Explain K-Nearest Neighbors and its working process.

K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for both classification and regression problems. It predicts the output of a new data point by looking at the K nearest data points in the training dataset.

**Working Process:**
1. Choose the value of K.
2. Calculate the distance between the new point and all training points (commonly Euclidean distance).
3. Select the K nearest neighbors.
4. For classification, choose the majority class among the neighbors.
5. For regression, take the average value of the neighbors.
6. Return the predicted result.

## Q2. Role of K in KNN

The value of K determines how many neighboring points are considered while making a prediction.

- **K too small (e.g., K = 1):** The model becomes sensitive to noise and may overfit.
- **K too large:** The model becomes too generalized and may underfit.
- A balanced value of K usually provides better performance.

## Q3. Differentiate between Classification and Regression using KNN

| Basis | KNN Classification | KNN Regression |
|---|---|---|
| Output | Category/Class label | Continuous value |
| Prediction Method | Majority voting | Average of neighbors |
| Example | Spam/Not Spam | House price prediction |

### Advantages of KNN
- Simple and easy to understand.
- No training phase is required.
- Works well for small datasets.
- Can solve both classification and regression problems.

### Disadvantages of KNN
- Computationally expensive for large datasets.
- Sensitive to irrelevant features and noise.
- Requires feature scaling.
- Choosing the correct K value can be difficult.

## Q4. KNN Classifier for Student Result Prediction


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

data = pd.DataFrame({
    'Study_Hours':[1,2,3,4,5,6],
    'Attendance':[50,55,60,70,75,85],
    'Result':['Fail','Fail','Fail','Pass','Pass','Pass']
})

le = LabelEncoder()
data['Result_Encoded'] = le.fit_transform(data['Result'])

X = data[['Study_Hours','Attendance']]
y = data['Result_Encoded']

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

prediction = model.predict([[4.5, 72]])
predicted_result = le.inverse_transform(prediction)[0]

train_accuracy = accuracy_score(y, model.predict(X))

print('Predicted Result:', predicted_result)
print('Accuracy Score:', train_accuracy)
```

## Q5. KNN Classifier for Fruit Classification


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, classification_report

data = pd.DataFrame({
    'Weight':[120,130,140,170,180,200],
    'Size':[6,6.5,7,8,8.5,9],
    'Fruit':['Apple','Apple','Apple','Orange','Orange','Orange']
})

le = LabelEncoder()
data['Fruit_Encoded'] = le.fit_transform(data['Fruit'])

X = data[['Weight','Size']]
y = data['Fruit_Encoded']

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

prediction = model.predict([[160, 7.5]])
predicted_fruit = le.inverse_transform(prediction)[0]

y_pred = model.predict(X)

print('Predicted Fruit:', predicted_fruit)
print('\nConfusion Matrix:')
print(confusion_matrix(y, y_pred))
print('\nClassification Report:')
print(classification_report(y, y_pred, target_names=le.classes_))
```
