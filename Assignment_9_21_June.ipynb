```python
# Q1 Evaluation Metrics

# MAE (Mean Absolute Error): Average of the absolute differences between actual and predicted values. Lower MAE means better accuracy.
# MSE (Mean Squared Error): Average of the squared differences between actual and predicted values. Larger errors are penalized more.
# RMSE (Root Mean Squared Error): Square root of MSE. It shows error in the same unit as the target variable. Lower RMSE indicates better performance.
# R² Score (Coefficient of Determination): Measures how well the model explains the variance in the data. Values range from 0 to 1, with values closer to 1 indicating a better fit.
```

```python
# Q2 Importance of Evaluation Metrics

# Help measure model performance.
# Allow comparison between different models.
# Detect underfitting and overfitting.
# Ensure reliable and accurate predictions.
```

```python
#Q3 Difference between Good Fit and Overfit Models

# A good fit model has low error values (low MAE, MSE, and RMSE) and a high R² score on both training and testing data, showing that it generalizes well.
# An overfit model performs extremely well on training data with very low errors and high R², but its performance becomes poor on testing data, with higher errors and lower R² score, indicating that it fails to generalize to new data.
```

```python
#Q4

import sys

print(sys.executable)
print(sys.version)

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Dataset
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([30, 40, 50, 60, 70])

# Train the model
model = LinearRegression()
model.fit(X, y)

# Predict outputs
y_pred = model.predict(X)

# Calculate errors
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)

# Display results
print("Predicted Marks:", y_pred)
print("MAE =", mae)
print("MSE =", mse)
print("RMSE =", rmse)
```

```python
# Import libraries
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Dataset
temperature = np.array([20, 25, 30, 35, 40]).reshape(-1, 1)
sales = np.array([100, 120, 150, 180, 200])

# Train the regression model
model = LinearRegression()
model.fit(temperature, sales)

# Predict sales
predicted_sales = model.predict(temperature)

# Calculate R² Score
r2 = r2_score(sales, predicted_sales)

# Display results
print("Slope (m):", model.coef_[0])
print("Intercept (c):", model.intercept_)

print("\nPredicted Sales:")
for temp, pred in zip(temperature.flatten(), predicted_sales):
    print(f"Temperature = {temp}°C, Predicted Sales = {pred:.2f}")

print("\nR² Score:", round(r2, 4))
```
